# 03 — Individual-Level Glucose-Insulin-Food ODE Model

Adapted from Kumnungkit et al. (2022). Simulates blood glucose and insulin dynamics for a single patient after meals.

**Key idea:** since PIMA is cross-sectional (no time-series data), this model cannot be *fitted* to PIMA directly. Instead, the ML classifier's predicted class (0/1) selects which literature-derived parameter regime (Normal vs Diabetic) to simulate, and a patient's real fasting glucose value from PIMA seeds the initial condition. See `src/ode_individual.py` docstring for the full reasoning.
